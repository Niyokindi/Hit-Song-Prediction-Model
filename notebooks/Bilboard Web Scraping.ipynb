```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
from datetime import datetime, timedelta
import time

# Define the start and end years
START_YEAR = 2000
END_YEAR = 2023

# Headers to mimic a real browser
HEADERS = {
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/85.0.4183.121 Safari/537.36"
}

# Initialize an empty list to store song data
all_songs = []

# Loop through each year and week
for year in range(START_YEAR, END_YEAR + 1):
    start_date = datetime(year, 1, 1)
    end_date = datetime(year, 12, 31)
    current_date = start_date
    
    print(f"Scraping Billboard Hot 100 for the year: {year}")
    
    while current_date <= end_date:
        date_string = current_date.strftime('%Y-%m-%d')
        url = f"https://www.billboard.com/charts/hot-100/{date_string}/"
        
        try:
            response = requests.get(url, headers=HEADERS)
            response.raise_for_status()
            
            soup = BeautifulSoup(response.content, 'html.parser')
            
            # Extract song data
            songs = soup.find_all('li', {'class': 'o-chart-results-list__item'})
            
            for song in songs:
                title_tag = song.find('h3', {'id': 'title-of-a-story'})
                artist_tag = song.find('span', {'class': 'a-no-trucate'})

                # Skip this song if either the title or artist does not exist
                if not title_tag or not artist_tag:
                    continue
                
                title = title_tag.text.strip()
                artist = artist_tag.text.strip()
                
                all_songs.append({
                    'Date': date_string,
                    'Title': title,
                    'Artist': artist,
                })
            
            print(f"Scraped data for week: {date_string}")
        except requests.exceptions.RequestException as e:
            print(f"Failed to retrieve data for {date_string}: {e}")
        
        # Pause to avoid rate-limiting
        time.sleep(2)
        
        # Move to the next week
        current_date += timedelta(days=7)

# Convert the list to a DataFrame
df = pd.DataFrame(all_songs)
df2 = df.drop_duplicates(subset=['Title', 'Artist'], keep='first')

output_file = 'billboard_hot100_2000_Jan.csv'
df2.to_csv(output_file, index=False)

print(f"Data scraping completed. File saved as {output_file}")


```
