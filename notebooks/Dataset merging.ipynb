```python
from rapidfuzz import fuzz, process

# Create a helper function for fuzzy matching
def is_in_billboard(song_title, artist_name, billboard):
    match = process.extractOne(
        (song_title, artist_name), 
        list(zip(billboard['Title'], billboard['Artist'])), 
        scorer=fuzz.token_sort_ratio
    )
    return match[1] >= 90  # Threshold for a match
```

```python
# Apply matching
songs['Hit'] = songs.apply(
    lambda row: is_in_billboard(row['track_name'], row['artist_name'], billboard), axis=1
).astype(int)

# Save the labeled dataset
songs.to_csv('songs.csv', index=False)
```
